```python
from pyspark.sql.functions import col
from pyspark.sql import SparkSession

spark = SparkSession.builder.getOrCreate()

# Load Gold data
gold_path = "/Volumes/azure-medallion-university-chapters/gold/university_chapters"
gold_df = spark.read.parquet(gold_path)

# Test: quarantined rows never in Gold
assert gold_df.filter(col("dq_status") == "INVALID_COORDINATES").count() == 0

# Test: warned rows appear in Gold
assert gold_df.filter(col("dq_status") == "WARNING").count() > 0
```
